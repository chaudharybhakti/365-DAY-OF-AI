Length of a String
We can find the length of a string using len() function.

```python
name="bhakti chaudhari"
print(len(name))
```

```python
name="harry sharma"
print(len(name))
```

```python
fruit = "Mango"
len1 = len(fruit)
print("Mango is a", len1, "letter word.")
```

```python
fruit = "Mango"
print("Mango is a", (len(fruit)), "letter word.")
```

String as an array
A string is essentially a sequence of characters also called an array. Thus we can access the elements of this array.

```python
dept="artificial intelligence"
print(dept[0:6])
```

```python
dept="artificial intelligence"
print(dept[:6])
```

```python
dept="artificial intelligence"
print(dept[4:6])
```

```python
dept="artificial intelligence"
print(dept[::6])
```

```python
dept="artificial intelligence"
print(dept[6::])
```

```python
dept="artificial intelligence"
print(dept[:])
```

```python
dept="artificial intelligence"
print(dept[3:-4])
```

```python
dept="artificial intelligence"
print(dept[:-5])
```

```python
dept="artificial intelligence"
print(dept[::-6])
```

```python
pie = "ApplePie"
print(pie[:5])
print(pie[6])	#returns character at specified index
```
